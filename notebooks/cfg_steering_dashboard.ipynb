## CFG Formula

Let:

- `v_none` be the model velocity with no background and no observations.
- `v_bg` be the velocity with background only.
- `v_obs` be the velocity with observations only.
- `v_both` be the velocity with background and observations together.

The recommended conservative mode is `background_delta`:

```text
v = v_none
  + cfg_background_scale  * (v_bg   - v_none)
  + cfg_observation_scale * (v_both - v_bg)
```

When both scales are `1.0`, this reduces to the normal fully conditioned prediction `v_both`.

The more aggressive mode is `independent`:

```text
v = v_none
  + cfg_background_scale  * (v_bg  - v_none)
  + cfg_observation_scale * (v_obs - v_none)
```

Use `cfg_background_scale` to tune background influence and `cfg_observation_scale` to tune observation influence.

````python
CONFIG_PATH = "config/experiments/evaluate_m2m_concat_conditioning_residual_balanced_2f.json"
RUN_DIR = "checkpoints/concat_conditioning/.../run_YYYYMMDD_HHMMSS"
CHECKPOINT_NAME = "ema_best_model.pth"

SPLIT = "valid"
CASE_INDEX = 23
NUM_TIMESTEPS = 25
METHOD = "euler"
SEED = 1234

OUTPUT_PATH = Path("outputs/cfg_steering_dashboard") / f"case_{CASE_INDEX:06d}.png"

STEERING_SETTINGS = [
    {
        "label": "baseline",
        "cfg_mode": "none",
        "cfg_background_scale": 1.0,
        "cfg_observation_scale": 1.0,
    },
    {
        "label": "obs x1.5",
        "cfg_mode": "background_delta",
        "cfg_background_scale": 1.0,
        "cfg_observation_scale": 1.5,
    },
]
````

## Render

````python
fig, metadata = make_steering_dashboard(
    config_path=CONFIG_PATH,
    run_dir=RUN_DIR,
    checkpoint_name=CHECKPOINT_NAME,
    steering_settings=STEERING_SETTINGS,
    split=SPLIT,
    case_index=CASE_INDEX,
    num_timesteps=NUM_TIMESTEPS,
    method=METHOD,
    seed=SEED,
    enforce_observations=False,
    obs_guidance_scale=0.0,
    output_path=OUTPUT_PATH,
    channels=None,
    panel_width=4.4,
    panel_height=3.2,
    dpi=180,
)

metadata
````
